##### Mini Project - Full Processing Pipeline

```python
import pandas as pd
import re
import string
import nltk

from sklearn.preprocessing import LabelEncoder
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

nltk.download("punkt")
nltk.download("stopwords")
```

```python
'''
Load the Dataset
'''

df = pd.read_csv(
    "train.txt",
    sep=";",
    names=["text", "emotions"]
)

print(df.head())
print("\nShape:", df.shape)
```

```python
'''
Encode Emotion Labels
'''

label_encoder = LabelEncoder()

df["label"] = label_encoder.fit_transform(df["emotions"])

print(dict(zip(
    label_encoder.classes_,
    label_encoder.transform(label_encoder.classes_)
)))
```

```python
'''
Complete Text Cleaning
'''

stop_words = set(stopwords.words("english"))

def clean_text(text):
    text = text.lower()
    text = text.translate(
        str.maketrans("", "", string.punctuation)
    )

    text = re.sub(r"\d+", "", text)
    text = text.encode("ascii", "ignore").decode("ascii")
    
    words = word_tokenize(text)
    words = [
        word for word in words
        if word not in stop_words
    ]
    return " ".join(words)

df["cleaned_text"] = df["text"].apply(clean_text)
print(df[["text", "cleaned_text"]].head())
```

```python
'''
Save Dataset and Display Emotion Counts
'''

df.to_csv(
    "cleaned_emotions.csv",
    index=False
)

print("Dataset saved as cleaned_emotions.csv")

print("\nEmotion Counts:")
print(df["emotions"].value_counts())
```
